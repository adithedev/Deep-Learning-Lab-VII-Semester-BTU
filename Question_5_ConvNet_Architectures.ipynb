## ConvNet Architectures

**Aim:** Implement famous ConvNet architectures: **AlexNet, ZFNet, VGG, C3D, GoogLeNet, ResNet and MobileNetV1.**


## 5.1 AlexNet

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

alexnet = Sequential([
    Conv2D(96,(11,11),strides=4,activation="relu",input_shape=(224,224,3)),
    MaxPooling2D((3,3),strides=2),
    Conv2D(256,(5,5),padding="same",activation="relu"),
    MaxPooling2D((3,3),strides=2),
    Conv2D(384,(3,3),padding="same",activation="relu"),
    Conv2D(384,(3,3),padding="same",activation="relu"),
    Conv2D(256,(3,3),padding="same",activation="relu"),
    MaxPooling2D((3,3),strides=2),
    Flatten(),
    Dense(4096,activation="relu"),
    Dense(4096,activation="relu"),
    Dense(1000,activation="softmax")
])
alexnet.summary()
```

## 5.2 ZFNet

```python
zfnet = Sequential([
    Conv2D(96,(7,7),strides=2,activation="relu",input_shape=(224,224,3)),
    MaxPooling2D((3,3),strides=2),
    Conv2D(256,(5,5),padding="same",activation="relu"),
    MaxPooling2D((3,3),strides=2),
    Conv2D(384,(3,3),padding="same",activation="relu"),
    Conv2D(384,(3,3),padding="same",activation="relu"),
    Conv2D(256,(3,3),padding="same",activation="relu"),
    MaxPooling2D((3,3),strides=2),
    Flatten(),
    Dense(4096,activation="relu"),
    Dense(4096,activation="relu"),
    Dense(1000,activation="softmax")
])
zfnet.summary()
```

## 5.3 VGG16

```python
vgg16 = tf.keras.applications.VGG16(
    weights=None, input_shape=(224,224,3), classes=1000
)
vgg16.summary()
```

## 5.4 C3D

```python
from tensorflow.keras.layers import Conv3D, MaxPooling3D

c3d = Sequential([
    Conv3D(64,(3,3,3),activation="relu",input_shape=(16,112,112,3)),
    MaxPooling3D((1,2,2)),
    Conv3D(128,(3,3,3),activation="relu"),
    MaxPooling3D((2,2,2)),
    Conv3D(256,(3,3,3),activation="relu"),
    MaxPooling3D((2,2,2)),
    Flatten(),
    Dense(4096,activation="relu"),
    Dense(101,activation="softmax")
])
c3d.summary()
```

## 5.5 GoogLeNet

TensorFlow/Keras provides the closely related **InceptionV3** implementation. It is used here as the available Keras implementation for the GoogLeNet/Inception family. It is not identical to the original GoogLeNet (Inception v1).


```python
googlenet_family = tf.keras.applications.InceptionV3(
    weights=None, input_shape=(299,299,3), classes=1000
)
googlenet_family.summary()
```

## 5.6 ResNet

```python
resnet50 = tf.keras.applications.ResNet50(
    weights=None, input_shape=(224,224,3), classes=1000
)
resnet50.summary()
```

## 5.7 MobileNetV1

```python
mobilenetv1 = tf.keras.applications.MobileNet(
    weights=None, input_shape=(224,224,3), classes=1000
)
mobilenetv1.summary()
```
