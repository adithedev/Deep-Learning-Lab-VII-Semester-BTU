## Custom Deep Learning Operations in PyTorch

**Aim:** Implement custom deep learning operations in PyTorch using gradient descent and the recursive chain rule (backpropagation), and demonstrate bias-variance tradeoff, regularization, output units (Linear, Softmax) and hidden units (Tanh, ReLU).


## A. Gradient Descent and Backpropagation

```python
import torch
import torch.nn as nn

torch.manual_seed(1)


# XOR dataset
X = torch.tensor([[0.,0.],[0.,1.],[1.,0.],[1.,1.]])
Y = torch.tensor([[0.],[1.],[1.],[0.]])

# Custom network: input -> Tanh hidden layer -> Sigmoid output
W1 = torch.randn(2, 4, requires_grad=True)
b1 = torch.zeros(4, requires_grad=True)
W2 = torch.randn(4, 1, requires_grad=True)
b2 = torch.zeros(1, requires_grad=True)

lr = 0.1

for epoch in range(5000):
    Z1 = X @ W1 + b1
    A1 = torch.tanh(Z1)              # hidden unit

    Z2 = A1 @ W2 + b2                # linear output before sigmoid
    A2 = torch.sigmoid(Z2)

    loss = torch.mean((A2 - Y)**2)
    loss.backward()                   # recursive chain rule

    with torch.no_grad():             # gradient descent update
        W1 -= lr * W1.grad
        b1 -= lr * b1.grad
        W2 -= lr * W2.grad
        b2 -= lr * b2.grad

        W1.grad.zero_()
        b1.grad.zero_()
        W2.grad.zero_()
        b2.grad.zero_()

print("Predictions:")
print(torch.round(A2, decimals=3).detach())
```

## B. Hidden and Output Units

```python
import torch
import torch.nn.functional as F

x = torch.tensor([-2., -1., 0., 1., 2.])

linear = x
relu = F.relu(x)
tanh = torch.tanh(x)

z = torch.tensor([1., 2., 3.])
softmax = F.softmax(z, dim=0)

print("Linear:", linear)
print("ReLU:", relu)
print("Tanh:", torch.round(tanh, decimals=3))
print("Softmax:", torch.round(softmax, decimals=3))
```

## C. Regularization

```python
# L2 regularization (weight decay) and Dropout in PyTorch
model = nn.Sequential(
    nn.Linear(784, 128),
    nn.ReLU(),
    nn.Dropout(0.3),
    nn.Linear(128, 64),
    nn.ReLU(),
    nn.Linear(64, 10)
)

optimizer = torch.optim.Adam(model.parameters(),
                             lr=0.001,
                             weight_decay=0.001)  # L2 regularization

print(model)
print("Dropout probability: 0.3")
print("L2 weight decay: 0.001")
```

## D. Bias-Variance Tradeoff

```python
# Simple demonstration:

for complexity in ["Low", "Medium", "High"]:
    print(complexity, "complexity model")

print("Low complexity  -> High bias, Low variance")
print("High complexity -> Low bias, High variance")
print("Regularization  -> Helps control variance")
```
